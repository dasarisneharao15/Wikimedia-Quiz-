```python
# ==========================================
# RELEVANCE-BASED WIKIMEDIA ARTICLE SEARCH
# ==========================================

topic = input("Enter a Wikipedia topic: ").strip()

if not topic:
    print("Please enter a topic.")
else:
    topic_lower = topic.lower()

    # Search articles containing the topic
    results = (
        pl.scan_parquet(DATA_PATH)
        .filter(
            pl.col("name")
            .str.to_lowercase()
            .str.contains(topic_lower)
        )
        .select(["name", "description", "abstract", "url"])
        .collect()
    )

    # Remove disambiguation pages
    results = results.filter(
        ~pl.col("description")
        .fill_null("")
        .str.to_lowercase()
        .str.contains("topics referred to")
    )

    # Give exact/specific titles priority
    results = results.with_columns(
        pl.when(
            pl.col("name").str.to_lowercase() == topic_lower
        )
        .then(0)
        .when(
            pl.col("name").str.to_lowercase().str.starts_with(topic_lower + " (")
        )
        .then(1)
        .otherwise(2)
        .alias("priority")
    )

    results = (
        results
        .sort("priority")
        .select(["name", "description", "abstract", "url"])
        .head(10)
    )

    if results.height == 0:
        print(f"\nNo suitable articles found for '{topic}'.")
    else:
        print(f"\nRelevant articles for '{topic}':")
        
        for i, name in enumerate(results["name"].to_list(), start=1):
            print(f"{i}. {name}")
```

```python
# ==========================================
# WIKIQUIZ - SMART TOPIC SEARCH
# ==========================================

topic = input("Enter a Wikipedia topic: ").strip()

if not topic:
    print("Please enter a topic.")
else:
    topic_lower = topic.lower()

    # First: search for an exact title
    exact = (
        pl.scan_parquet(DATA_PATH)
        .filter(
            pl.col("name").str.to_lowercase() == topic_lower
        )
        .select(["name", "description", "abstract", "url"])
        .collect()
    )

    # If exact result is a disambiguation page, don't use it
    if exact.height > 0:
        exact_description = str(exact["description"][0]).lower()

        if "topics referred to" not in exact_description:
            results = exact
        else:
            results = pl.DataFrame()
    else:
        results = pl.DataFrame()

    # If exact article isn't suitable, search related titles
    if results.height == 0:

        results = (
            pl.scan_parquet(DATA_PATH)
            .filter(
                pl.col("name")
                .str.to_lowercase()
                .str.contains(topic_lower)
            )
            .select(["name", "description", "abstract", "url"])
            .collect()
        )

        # Remove disambiguation pages
        results = results.filter(
            ~pl.col("description")
            .fill_null("")
            .str.to_lowercase()
            .str.contains("topics referred to")
        )

        # Give useful article types higher priority
        results = results.with_columns(
            pl.when(
                pl.col("name")
                .str.to_lowercase()
                .str.contains("programming language")
            )
            .then(0)
            .when(
                pl.col("description")
                .fill_null("")
                .str.to_lowercase()
                .contains("programming language")
            )
            .then(1)
            .when(
                pl.col("name")
                .str.to_lowercase()
                .str.contains(topic_lower + " \\(")
            )
            .then(2)
            .otherwise(3)
            .alias("priority")
        )

        results = (
            results
            .sort("priority")
            .select(["name", "description", "abstract", "url"])
            .head(10)
        )

    if results.height == 0:
        print(f"\nNo suitable article found for '{topic}'.")
    else:
        print(f"\nArticles found for '{topic}':\n")

        for i, name in enumerate(results["name"].to_list(), start=1):
            print(f"{i}. {name}")
```

```python
# ============================================================
#                 WIKIQUIZ - UI VERSION
#          Interactive Wikimedia Quiz Generator
# ============================================================

import polars as pl
import random
import re
import streamlit as st


# ------------------------------------------------------------
# DATASET PATH
# ------------------------------------------------------------

DATA_PATH = "/kaggle/input/datasets/wikimedia-foundation/wikipedia-structured-contents/enwiki/data/*.parquet"


# ------------------------------------------------------------
# PAGE CONFIGURATION
# ------------------------------------------------------------

st.set_page_config(
    page_title="WikiQuiz",
    page_icon="🧠",
    layout="centered"
)


# ------------------------------------------------------------
# CUSTOM CSS
# ------------------------------------------------------------

st.markdown("""
<style>

.main {
    background-color: #f7f9fc;
}

.title {
    text-align: center;
    font-size: 45px;
    font-weight: bold;
    margin-bottom: 5px;
}

.subtitle {
    text-align: center;
    font-size: 18px;
    color: #666;
    margin-bottom: 30px;
}

.article-card {
    padding: 20px;
    border-radius: 15px;
    background-color: #ffffff;
    border: 1px solid #dddddd;
    margin-bottom: 15px;
}

.result-card {
    padding: 30px;
    border-radius: 20px;
    text-align: center;
    background-color: #ffffff;
    border: 1px solid #dddddd;
}

.score {
    font-size: 50px;
    font-weight: bold;
}

</style>
""", unsafe_allow_html=True)


# ------------------------------------------------------------
# TITLE
# ------------------------------------------------------------

st.markdown(
    '<div class="title">🧠 WikiQuiz</div>',
    unsafe_allow_html=True
)

st.markdown(
    '<div class="subtitle">'
    'Interactive Wikimedia Quiz Generator'
    '</div>',
    unsafe_allow_html=True
)


# ------------------------------------------------------------
# SESSION STATE
# ------------------------------------------------------------

if "results" not in st.session_state:
    st.session_state.results = None

if "selected_article" not in st.session_state:
    st.session_state.selected_article = None

if "questions" not in st.session_state:
    st.session_state.questions = []

if "current_question" not in st.session_state:
    st.session_state.current_question = 0

if "score" not in st.session_state:
    st.session_state.score = 0

if "quiz_started" not in st.session_state:
    st.session_state.quiz_started = False

if "answered" not in st.session_state:
    st.session_state.answered = False

if "last_answer_correct" not in st.session_state:
    st.session_state.last_answer_correct = False


# ------------------------------------------------------------
# SEARCH TOPIC
# ------------------------------------------------------------

st.subheader("🔎 Search Wikipedia")

topic = st.text_input(
    "Enter a Wikipedia topic",
    placeholder="Example: India, Python, Machine Learning"
)


if st.button("Search Articles", use_container_width=True):

    if not topic.strip():

        st.warning("Please enter a topic.")

    else:

        topic_lower = topic.strip().lower()

        with st.spinner("Searching Wikimedia dataset..."):

            results = (
                pl.scan_parquet(DATA_PATH)
                .filter(
                    pl.col("name")
                    .str.to_lowercase()
                    .str.contains(
                        topic_lower,
                        literal=True
                    )
                )
                .select([
                    "name",
                    "description",
                    "abstract",
                    "url"
                ])
                .collect()
            )


        # Remove disambiguation pages

        results = results.filter(
            ~pl.col("description")
            .fill_null("")
            .str.to_lowercase()
            .str.contains(
                "topics referred to",
                literal=True
            )
        )


        # Remove duplicate article names

        results = results.unique(
            subset=["name"],
            keep="first"
        )


        # Rank articles

        names = results["name"].to_list()

        descriptions = (
            results["description"]
            .fill_null("")
            .to_list()
        )

        priority = []


        for name, description in zip(
            names,
            descriptions
        ):

            name_lower = name.lower()
            description_lower = description.lower()

            if name_lower == topic_lower:

                priority.append(0)

            elif name_lower.startswith(
                topic_lower + " ("
            ):

                priority.append(1)

            elif (
                "programming language"
                in name_lower
                or
                "programming language"
                in description_lower
            ):

                priority.append(2)

            else:

                priority.append(3)


        results = results.with_columns(
            pl.Series(
                "priority",
                priority
            )
        )


        results = (
            results
            .sort("priority")
            .drop("priority")
            .head(10)
        )


        st.session_state.results = results
        st.session_state.selected_article = None
        st.session_state.questions = []
        st.session_state.quiz_started = False
        st.session_state.current_question = 0
        st.session_state.score = 0
        st.session_state.answered = False


# ------------------------------------------------------------
# DISPLAY SEARCH RESULTS
# ------------------------------------------------------------

if st.session_state.results is not None:

    results = st.session_state.results

    if results.height == 0:

        st.error(
            f"No suitable articles found for '{topic}'."
        )

    else:

        st.subheader("📚 Articles Found")

        article_names = results["name"].to_list()

        selected_name = st.radio(
            "Select an article:",
            article_names
        )


        if st.button(
            "Select Article",
            use_container_width=True
        ):

            selected_rows = results.filter(
                pl.col("name") == selected_name
            )

            st.session_state.selected_article = (
                selected_rows.row(
                    0,
                    named=True
                )
            )


            # Reset quiz

            st.session_state.questions = []
            st.session_state.current_question = 0
            st.session_state.score = 0
            st.session_state.quiz_started = False
            st.session_state.answered = False


# ------------------------------------------------------------
# DISPLAY SELECTED ARTICLE
# ------------------------------------------------------------

if st.session_state.selected_article is not None:

    article = st.session_state.selected_article

    article_title = article["name"]

    article_description = article["description"]

    article_abstract = article["abstract"]

    article_url = article["url"]


    st.markdown("---")

    st.subheader("📖 Selected Article")


    st.markdown(
        f"""
        <div class="article-card">

        <h2>{article_title}</h2>

        <p>
        <b>Description:</b>
        {article_description}
        </p>

        <p>
        Information extracted from the
        Wikimedia Structured Contents dataset.
        </p>

        </div>
        """,
        unsafe_allow_html=True
    )


    st.markdown(
        f"**Source:** {article_url}"
    )


    # --------------------------------------------------------
    # GENERATE QUIZ
    # --------------------------------------------------------

    if st.button(
        "🎯 Generate Quiz",
        use_container_width=True
    ):

        if article_abstract is None:

            article_abstract = ""


        text = str(
            article_abstract
        ).strip()


        sentences = re.split(
            r'(?<=[.!?])\s+',
            text
        )


        useful_sentences = []


        for sentence in sentences:

            sentence = sentence.strip()

            if (
                len(sentence) >= 40
                and len(sentence) <= 250
            ):

                useful_sentences.append(
                    sentence
                )


        questions = []


        # ----------------------------------------------------
        # QUESTION 1
        # ----------------------------------------------------

        if (
            article_description is not None
            and
            str(article_description).strip() != ""
        ):

            description = str(
                article_description
            ).strip()


            questions.append({

                "question":
                    f"What best describes "
                    f"'{article_title}'?",

                "options": [

                    description,

                    "A type of database system",

                    "A computer hardware device",

                    "A fictional character"

                ],

                "answer": description

            })


        # ----------------------------------------------------
        # QUESTIONS FROM ARTICLE
        # ----------------------------------------------------

        alternative_questions = [

            "Which statement is supported by the selected article?",

            "Which of the following is mentioned in the article?",

            "Which statement is correct according to the article?",

            "According to the Wikimedia article, which statement is correct?"

        ]


        wrong_options = [

            [
                "The article contains no information about this topic.",
                "The topic is only related to sports.",
                "The topic is a database command."
            ],

            [
                "The article contains only numerical data.",
                "The topic is exclusively a programming error.",
                "The article states that this topic does not exist."
            ],

            [
                "The article is unrelated to the selected topic.",
                "The article describes only computer hardware.",
                "The article contains no factual information."
            ],

            [
                "This information is not mentioned in the article.",
                "The article is about database administration.",
                "The article describes only computer hardware."
            ]

        ]


        for i, fact in enumerate(
            useful_sentences[:4]
        ):

            wrongs = wrong_options[
                min(
                    i,
                    len(wrong_options) - 1
                )
            ]


            questions.append({

                "question":
                    alternative_questions[
                        min(
                            i,
                            len(alternative_questions) - 1
                        )
                    ],

                "options": [
                    fact
                ] + wrongs,

                "answer": fact

            })


        # Maximum 5 questions

        questions = questions[:5]


        # Shuffle options

        for question in questions:

            random.shuffle(
                question["options"]
            )


        st.session_state.questions = questions

        st.session_state.current_question = 0

        st.session_state.score = 0

        st.session_state.quiz_started = True

        st.session_state.answered = False


# ------------------------------------------------------------
# QUIZ SECTION
# ------------------------------------------------------------

if (
    st.session_state.quiz_started
    and
    len(st.session_state.questions) > 0
):

    questions = st.session_state.questions

    current = st.session_state.current_question


    st.markdown("---")

    st.subheader("📝 Quiz")


    # --------------------------------------------------------
    # RESULT
    # --------------------------------------------------------

    if current >= len(questions):

        score = st.session_state.score

        total = len(questions)

        percentage = (
            score / total
        ) * 100


        st.markdown(
            '<div class="result-card">',
            unsafe_allow_html=True
        )


        st.markdown(
            "## 🎉 Quiz Completed!"
        )


        st.markdown(
            f'<div class="score">'
            f'{score}/{total}'
            f'</div>',
            unsafe_allow_html=True
        )


        st.markdown(
            f"### {percentage:.0f}%"
        )


        if score == total:

            st.success(
                "Excellent! 🎉"
            )

        elif score >= 3:

            st.info(
                "Good job! 👍"
            )

        else:

            st.warning(
                "Keep learning and try again!"
            )


        st.markdown(
            "</div>",
            unsafe_allow_html=True
        )


        if st.button(
            "🔄 Play Again",
            use_container_width=True
        ):

            st.session_state.quiz_started = False

            st.session_state.current_question = 0

            st.session_state.score = 0

            st.session_state.answered = False

            st.rerun()


    # --------------------------------------------------------
    # CURRENT QUESTION
    # --------------------------------------------------------

    else:

        question = questions[current]


        st.progress(
            (current + 1) / len(questions)
        )


        st.write(
            f"**Question {current + 1} "
            f"of {len(questions)}**"
        )


        st.markdown(
            f"### {question['question']}"
        )


        selected_answer = st.radio(
            "Choose your answer:",
            question["options"],
            key=f"question_{current}"
        )


        # ----------------------------------------------------
        # SUBMIT ANSWER
        # ----------------------------------------------------

        if not st.session_state.answered:

            if st.button(
                "Submit Answer",
                use_container_width=True
            ):

                if (
                    selected_answer
                    == question["answer"]
                ):

                    st.session_state.score += 1

                    st.session_state.last_answer_correct = True

                    st.success(
                        "Correct! ✓"
                    )

                else:

                    st.session_state.last_answer_correct = False

                    st.error(
                        "Wrong!"
                    )

                    st.info(
                        "Correct answer: "
                        + question["answer"]
                    )


                st.session_state.answered = True

                st.rerun()


        # ----------------------------------------------------
        # NEXT QUESTION
        # ----------------------------------------------------

        else:

            if current + 1 < len(questions):

                if st.button(
                    "Next Question →",
                    use_container_width=True
                ):

                    st.session_state.current_question += 1

                    st.session_state.answered = False

                    st.rerun()

            else:

                if st.button(
                    "View Result →",
                    use_container_width=True
                ):

                    st.session_state.current_question += 1

                    st.rerun()


# ------------------------------------------------------------
# FOOTER
# ------------------------------------------------------------

st.markdown("---")

st.caption(
    "WikiQuiz • Powered by Wikimedia Structured Contents"
)
```

```python
# ============================================================
#                  WIKIQUIZ - INTERACTIVE UI
#          Wikimedia Structured Contents Dataset
# ============================================================

import polars as pl
import random
import re
import ipywidgets as widgets
from IPython.display import display, clear_output


# ============================================================
# DATASET PATH
# ============================================================

DATA_PATH = "/kaggle/input/datasets/wikimedia-foundation/wikipedia-structured-contents/enwiki/data/*.parquet"


# ============================================================
# UI STYLING
# ============================================================

title = widgets.HTML("""
<div style="
    background: linear-gradient(135deg, #4f46e5, #7c3aed);
    padding: 28px;
    border-radius: 18px;
    color: white;
    text-align: center;
    margin-bottom: 20px;
">
    <h1 style="margin:0;font-size:34px;">🧠 WikiQuiz</h1>
    <p style="margin:8px 0 0;font-size:17px;">
        Interactive Wikimedia Quiz Generator
    </p>
</div>
""")


# ============================================================
# TOPIC INPUT
# ============================================================

topic_input = widgets.Text(
    placeholder="Enter a Wikipedia topic...",
    description="Topic:",
    layout=widgets.Layout(width="70%")
)

search_button = widgets.Button(
    description="🔍 Search",
    button_style="primary",
    layout=widgets.Layout(width="120px")
)

search_output = widgets.Output()


# ============================================================
# ARTICLE SELECTION
# ============================================================

article_dropdown = widgets.Dropdown(
    options=[],
    description="Article:",
    layout=widgets.Layout(width="80%"),
    disabled=True
)

select_button = widgets.Button(
    description="📖 Select Article",
    button_style="success",
    disabled=True
)

article_output = widgets.Output()


# ============================================================
# QUIZ AREA
# ============================================================

quiz_output = widgets.Output()

start_quiz_button = widgets.Button(
    description="▶ Start Quiz",
    button_style="primary",
    disabled=True
)


# ============================================================
# GLOBAL VARIABLES
# ============================================================

results = None
selected_article = None
questions = []
current_question = 0
score = 0


# ============================================================
# GENERATE QUESTIONS
# ============================================================

def generate_questions(article):

    article_title = article["name"]
    article_description = article["description"]
    article_abstract = article["abstract"]

    if article_abstract is None:
        article_abstract = ""

    text = str(article_abstract).strip()

    # Split abstract into sentences
    sentences = re.split(
        r'(?<=[.!?])\s+',
        text
    )

    # Keep useful sentences
    useful_sentences = []

    for sentence in sentences:

        sentence = sentence.strip()

        if 40 <= len(sentence) <= 250:
            useful_sentences.append(sentence)

    generated = []


    # --------------------------------------------------------
    # Q1 - DESCRIPTION
    # --------------------------------------------------------

    if (
        article_description is not None
        and str(article_description).strip() != ""
    ):

        description = str(article_description).strip()

        generated.append({
            "question":
                f"What best describes '{article_title}'?",

            "options": [
                description,
                "A type of database system",
                "A computer hardware device",
                "A fictional character"
            ],

            "answer": description
        })


    # --------------------------------------------------------
    # Q2 - FIRST FACT
    # --------------------------------------------------------

    if len(useful_sentences) >= 1:

        fact = useful_sentences[0]

        generated.append({
            "question":
                "Which statement is supported by the selected article?",

            "options": [
                fact,
                "The article contains no information about this topic.",
                "The topic is only related to sports.",
                "The topic is a database command."
            ],

            "answer": fact
        })


    # --------------------------------------------------------
    # Q3 - SECOND FACT
    # --------------------------------------------------------

    if len(useful_sentences) >= 2:

        fact = useful_sentences[1]

        generated.append({
            "question":
                "Which of the following is mentioned in the article?",

            "options": [
                fact,
                "The article states that this topic does not exist.",
                "The topic is exclusively a programming error.",
                "The article contains only numerical data."
            ],

            "answer": fact
        })


    # --------------------------------------------------------
    # Q4 - THIRD FACT
    # --------------------------------------------------------

    if len(useful_sentences) >= 3:

        fact = useful_sentences[2]

        generated.append({
            "question":
                "Which statement is correct according to the article?",

            "options": [
                fact,
                "The article is unrelated to the selected topic.",
                "The article describes only computer hardware.",
                "The article contains no factual information."
            ],

            "answer": fact
        })


    # --------------------------------------------------------
    # Q5 - FOURTH FACT
    # --------------------------------------------------------

    if len(useful_sentences) >= 4:

        fact = useful_sentences[3]

        generated.append({
            "question":
                "According to the Wikimedia article, which statement is correct?",

            "options": [
                fact,
                "This information is not mentioned in the article.",
                "The article is about database administration.",
                "The article describes only computer hardware."
            ],

            "answer": fact
        })


    # Maximum 5 questions
    generated = generated[:5]

    # Shuffle options
    for q in generated:
        random.shuffle(q["options"])

    return generated


# ============================================================
# SEARCH FUNCTION
# ============================================================

def search_articles(button):

    global results

    with search_output:

        clear_output()

        topic = topic_input.value.strip()

        if not topic:

            display(widgets.HTML("""
            <div style="
                padding:12px;
                background:#fff3cd;
                border-radius:10px;
            ">
            ⚠️ Please enter a topic first.
            </div>
            """))

            return

        topic_lower = topic.lower()

        display(widgets.HTML(
            "<b>🔎 Searching Wikimedia dataset...</b>"
        ))

        # ----------------------------------------------------
        # SEARCH
        # ----------------------------------------------------

        results = (
            pl.scan_parquet(DATA_PATH)
            .filter(
                pl.col("name")
                .str.to_lowercase()
                .str.contains(
                    topic_lower,
                    literal=True
                )
            )
            .select([
                "name",
                "description",
                "abstract",
                "url"
            ])
            .collect()
        )


        # ----------------------------------------------------
        # REMOVE DISAMBIGUATION
        # ----------------------------------------------------

        results = results.filter(
            ~pl.col("description")
            .fill_null("")
            .str.to_lowercase()
            .str.contains(
                "topics referred to",
                literal=True
            )
        )


        # ----------------------------------------------------
        # REMOVE DUPLICATES
        # ----------------------------------------------------

        results = results.unique(
            subset=["name"],
            keep="first"
        )


        # ----------------------------------------------------
        # RANK RESULTS
        # ----------------------------------------------------

        names = results["name"].to_list()

        descriptions = (
            results["description"]
            .fill_null("")
            .to_list()
        )

        priority = []

        for name, description in zip(
            names,
            descriptions
        ):

            name_lower = name.lower()

            description_lower = description.lower()

            if name_lower == topic_lower:

                priority.append(0)

            elif name_lower.startswith(
                topic_lower + " ("
            ):

                priority.append(1)

            elif (
                "programming language" in name_lower
                or
                "programming language" in description_lower
            ):

                priority.append(2)

            else:

                priority.append(3)


        results = results.with_columns(
            pl.Series(
                "priority",
                priority
            )
        )


        results = (
            results
            .sort("priority")
            .drop("priority")
            .head(10)
        )


        # ----------------------------------------------------
        # DISPLAY RESULTS
        # ----------------------------------------------------

        if results.height == 0:

            display(widgets.HTML(f"""
            <div style="
                padding:15px;
                background:#fee2e2;
                border-radius:10px;
            ">
            ❌ No suitable articles found for
            <b>{topic}</b>.
            </div>
            """))

            article_dropdown.options = []

            article_dropdown.disabled = True

            select_button.disabled = True

            return


        # Create dropdown options

        options = []

        for i, name in enumerate(
            results["name"].to_list()
        ):

            options.append(
                (f"{i+1}. {name}", i)
            )


        article_dropdown.options = options

        article_dropdown.disabled = False

        select_button.disabled = False


        display(widgets.HTML(f"""
        <div style="
            padding:15px;
            background:#dcfce7;
            border-radius:10px;
            margin-top:10px;
        ">
        ✅ Found <b>{results.height}</b> suitable articles.
        <br>
        Select one from the dropdown below.
        </div>
        """))


# ============================================================
# ARTICLE SELECTION FUNCTION
# ============================================================

def select_article(button):

    global selected_article

    with article_output:

        clear_output()

        if results is None or results.height == 0:

            return

        index = article_dropdown.value

        selected_article = results.row(
            index,
            named=True
        )


        title_text = selected_article["name"]

        description = selected_article["description"]

        abstract = selected_article["abstract"]

        url = selected_article["url"]


        display(widgets.HTML(f"""
        <div style="
            background:#f8fafc;
            border:1px solid #e2e8f0;
            padding:22px;
            border-radius:15px;
            margin-top:15px;
        ">

        <h2 style="margin-top:0;">
        📖 {title_text}
        </h2>

        <p>
        <b>Description:</b>
        {description}
        </p>

        <p>
        <b>Source:</b>
        Wikimedia Structured Contents Dataset
        </p>

        <p>
        <b>Article URL:</b>
        {url}
        </p>

        <hr>

        <p>
        <b>Article Information:</b>
        </p>

        <p style="line-height:1.6;">
        {str(abstract)[:1200]}
        </p>

        </div>
        """))


        # Generate quiz
        generated = generate_questions(
            selected_article
        )


        if len(generated) == 0:

            display(widgets.HTML("""
            <div style="
                margin-top:15px;
                padding:15px;
                background:#fee2e2;
                border-radius:10px;
            ">
            ❌ Not enough information to generate
            questions for this article.
            </div>
            """))

            start_quiz_button.disabled = True

            return


        # Store generated questions
        global questions

        questions = generated


        display(widgets.HTML(f"""
        <div style="
            margin-top:15px;
            padding:15px;
            background:#eef2ff;
            border-radius:10px;
            text-align:center;
        ">
        <h3 style="margin:0;">
        🎯 {len(questions)} Questions Generated
        </h3>

        <p>
        Questions were generated from the selected
        Wikimedia article information.
        </p>

        </div>
        """))


        start_quiz_button.disabled = False


# ============================================================
# START QUIZ
# ============================================================

def start_quiz(button):

    global current_question
    global score

    current_question = 0
    score = 0

    show_question()


# ============================================================
# SHOW QUESTION
# ============================================================

def show_question():

    with quiz_output:

        clear_output()

        if current_question >= len(questions):

            show_result()

            return


        q = questions[current_question]


        # ----------------------------------------------------
        # PROGRESS
        # ----------------------------------------------------

        progress = (
            current_question + 1
        ) / len(questions) * 100


        display(widgets.HTML(f"""
        <div style="margin-top:20px;">

        <div style="
            height:10px;
            background:#e5e7eb;
            border-radius:10px;
            overflow:hidden;
        ">

        <div style="
            width:{progress}%;
            height:100%;
            background:#6366f1;
        "></div>

        </div>

        <p style="text-align:right;">
        Question {current_question + 1}
        of {len(questions)}
        </p>

        </div>
        """))


        # ----------------------------------------------------
        # QUESTION
        # ----------------------------------------------------

        question_html = widgets.HTML(f"""
        <div style="
            background:#ffffff;
            border:1px solid #e5e7eb;
            padding:20px;
            border-radius:15px;
            box-shadow:0 3px 10px rgba(0,0,0,0.05);
        ">

        <h3>
        Q{current_question + 1}. {q["question"]}
        </h3>

        </div>
        """)

        display(question_html)


        # ----------------------------------------------------
        # OPTIONS
        # ----------------------------------------------------

        option_buttons = []

        for i, option in enumerate(
            q["options"]
        ):

            btn = widgets.Button(
                description=f"{i+1}. {option}",
                layout=widgets.Layout(
                    width="95%",
                    min_height="55px"
                ),
                button_style=""
            )


            def make_handler(
                selected_option,
                selected_button
            ):

                def handler(b):

                    check_answer(
                        selected_option,
                        selected_button,
                        option_buttons
                    )

                return handler


            btn.on_click(
                make_handler(
                    option,
                    btn
                )
            )

            option_buttons.append(btn)

            display(btn)


# ============================================================
# CHECK ANSWER
# ============================================================

def check_answer(
    selected_option,
    selected_button,
    all_buttons
):

    global score
    global current_question


    correct_answer = questions[
        current_question
    ]["answer"]


    # Disable buttons
    for btn in all_buttons:
        btn.disabled = True


    # Correct
    if selected_option == correct_answer:

        score += 1

        selected_button.button_style = "success"

        display(widgets.HTML("""
        <div style="
            margin-top:15px;
            padding:12px;
            background:#dcfce7;
            border-radius:10px;
            color:#166534;
        ">
        ✅ Correct!
        </div>
        """))


    # Wrong
    else:

        selected_button.button_style = "danger"

        display(widgets.HTML(f"""
        <div style="
            margin-top:15px;
            padding:12px;
            background:#fee2e2;
            border-radius:10px;
            color:#991b1b;
        ">
        ❌ Wrong!<br>
        <b>Correct answer:</b>
        {correct_answer}
        </div>
        """))


    # Next button

    next_button = widgets.Button(
        description="➡ Next Question",
        button_style="primary",
        layout=widgets.Layout(
            width="200px"
        )
    )


    def next_question(b):

        global current_question

        current_question += 1

        show_question()


    next_button.on_click(
        next_question
    )

    display(next_button)


# ============================================================
# SHOW RESULT
# ============================================================

def show_result():

    percentage = (
        score / len(questions)
    ) * 100


    if percentage == 100:

        message = "🎉 Excellent! Perfect Score!"

    elif percentage >= 60:

        message = "👏 Good job! Keep learning!"

    else:

        message = "📚 Keep learning and try again!"


    with quiz_output:

        clear_output()


        display(widgets.HTML(f"""
        <div style="
            margin-top:25px;
            padding:35px;
            background:linear-gradient(
                135deg,
                #eef2ff,
                #f5f3ff
            );
            border-radius:20px;
            text-align:center;
        ">

        <h1>🏆 Quiz Completed!</h1>

        <h2>
        {selected_article["name"]}
        </h2>

        <div style="
            font-size:48px;
            font-weight:bold;
            margin:20px;
        ">
        {score}/{len(questions)}
        </div>

        <h2>
        {percentage:.0f}%
        </h2>

        <h3>
        {message}
        </h3>

        </div>
        """))


        restart_button = widgets.Button(
            description="🔄 Try Again",
            button_style="primary",
            layout=widgets.Layout(
                width="180px"
            )
        )


        def restart(b):

            global current_question
            global score

            current_question = 0
            score = 0

            show_question()


        restart_button.on_click(
            restart
        )


        display(restart_button)


# ============================================================
# CONNECT BUTTONS
# ============================================================

search_button.on_click(
    search_articles
)

select_button.on_click(
    select_article
)

start_quiz_button.on_click(
    start_quiz
)


# ============================================================
# DISPLAY COMPLETE UI
# ============================================================

display(title)

display(
    widgets.HTML("""
    <h3>🔎 Step 1 — Search for a topic</h3>
    """)
)

display(
    widgets.HBox([
        topic_input,
        search_button
    ])
)

display(search_output)

display(
    widgets.HTML("""
    <h3 style="margin-top:25px;">
    📚 Step 2 — Select an article
    </h3>
    """)
)

display(article_dropdown)

display(select_button)

display(article_output)

display(start_quiz_button)

display(quiz_output)
```
